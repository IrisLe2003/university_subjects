# CLEAN DATA

```python
#Kiểm tra giá trị NULL của từng cột
missing_data = df.isnull().sum()
missing_data.to_frame().T
```

```python
#Loại bỏ cột có giá trị NULL > 70%
#đặt ngưỡng = 0.7
threshold = 0.7

#tính tỉ lệ các giá trị NULL của từng cột
missing_percentages = df.isnull().mean()
missing_percentages = missing_percentages.to_frame().T
missing_percentages

#lọc các cột có giá trị > 0.7
columns_to_drop = missing_percentages.columns[missing_percentages.iloc[0] > threshold]
columns_to_drop

#xóa các cột có giá trị > 0.7
df = df.drop(columns=columns_to_drop)
df.head()
df.shape
```

# ANALYZE and VISUALIZE DATA

## Chủ nghĩa khủng bố đang ngày càng trở nên nghiêm trọng hơn

### Đa số các cuộc khủng bố xảy ra đều thành công

```python
#Vẽ biểu đồ thể hiện tổng số vụ khủng bố xảy ra và tổng số vụ khủng bố thành công theo từng năm

#danh sách các năm từ 1970-2017 (trừ năm 1993 không có thông tin trong tập dữ liệu)
year = df['iyear'].unique()

#số lượng các vụ khủng bố xảy ra trong mỗi năm theo thứ tự năm tăng dần
total_attack = df['iyear'].value_counts(dropna=False).sort_index()

#số lượng các vụ khủng bố được thực hiện thành công trong mỗi năm theo thứ tự năm tăng dần
success_attack = df[df['success']==1].groupby('iyear').size()

#Tạo một DataFrame 
plot_data = pd.DataFrame({'Total Attacks': total_attack,
                          'Successful Attacks': success_attack
                         }).reset_index()
plot_data.columns = ['Year', 'Total Attacks', 'Successful Attacks']

#tạo biểu đồ đường
plt.figure(figsize=(18, 10))
sns.lineplot(x='Year', y='Total Attacks', data=plot_data, color='black', alpha=0.5, label='Total Attacks')
sns.lineplot(x='Year', y='Successful Attacks', data=plot_data, color='red', alpha=0.5, label='Successful Attacks')

#thêm thông tin về biểu đồ
plt.xlabel('Year')
plt.xticks(rotation=45)
plt.ylabel('Number of Attacks')
plt.title('Total Attacks vs Successful Attacks by Year')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
```

__Note__: Một khi các phần tử khủng bố tiến hành các cuộc tấn công, thì gần như toàn bộ chúng đều thành công. Đặc biệt, kể từ thế kỉ 21 (năm 2000 đổ đi), ta thấy chủ nghĩa khủng bố ngày càng gia tăng.

```python
#Vẽ biểu đồ cột thể hiện tổng số lượng các cuộc khủng bố theo năm

#tạo biểu đồ cột
plt.figure(figsize = (18,10))
sns.barplot(x = year,
            y = total_attack,
            palette = 'rocket_r')

#thêm thông tin về biểu đồ
plt.xlabel('Year')
plt.xticks(rotation = 45)
plt.ylabel('Number of Attacks')
plt.title('Number of Attacks by Year')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.show()
```

__Note__: Năm 2014 xảy ra nhiều cuộc tấn công khủng bố nhất với gần 17000 vụ xảy ra trong một năm. 

### Sự an toàn của người dân bị đe dọa

```python
#Vẽ biểu đồ cột chồng thể hiện số người bị thương, số người tử vong và tổng số thương vong

#số lượng người bị thương trong mỗi năm
wound_year = df.groupby('iyear')['nwound'].sum()

#số lượng người tử vong trong mỗi năm
kill_year = df.groupby('iyear')['nkill'].sum()

#tạo DataFrame mới chứa cả hai loại dữ liệu
plot_data = pd.DataFrame({'Kill': kill_year,
                          'Wound': wound_year
                         }).reset_index()
plot_data.columns = ['Year', 'Kill', 'Wound']

#tính tổng số nạn nhân và phần trăm
plot_data['Total'] = plot_data['Kill'] + plot_data['Wound']
plot_data['Kill_Percent'] = (plot_data['Kill'] / plot_data['Total']) * 100
plot_data['Wound_Percent'] = (plot_data['Wound'] / plot_data['Total']) * 100
plot_data = plot_data.sort_values('Year')

#tạo biểu đồ
plt.figure(figsize=(18, 12))
#vẽ cột thể hiện số người tử vong
bars1 = plt.barh(plot_data['Year'], plot_data['Kill'], color='grey', alpha=0.7, label='Killed')
#vẽ cột thể hiện số người bị thương, chồng lên cột số người tử vong
bars2 = plt.barh(plot_data['Year'], plot_data['Wound'], left=plot_data['Kill'], color='red', alpha=0.7, label='Wounded')

#thêm thông tin về biểu đồ
plt.xlabel('Number of Casualties', fontsize=12)
plt.ylabel('Attack Year', fontsize=12)
plt.yticks(plot_data['Year'])
plt.title('Killed and Wounded by Year in Terrorist Attacks', fontsize=16)
plt.legend(fontsize=10)
plt.grid(axis='x', linestyle='--', alpha=0.3)

#thêm phần trăm vào trong các thanh
def add_percentages(bars, percentages, xpos='center'):
    for bar, percentage in zip(bars, percentages):
        width = bar.get_width()
        if width > 0:
            x = bar.get_x() + (width/2 if xpos == 'center' else width)
            plt.text(x, bar.get_y() + bar.get_height()/2.,
                     f'{percentage:.1f}%',
                     ha='center', va='center', color='black')
add_percentages(bars1, plot_data['Kill_Percent'], xpos='right')
add_percentages(bars2, plot_data['Wound_Percent'], xpos='right')

plt.tight_layout()
plt.show()
```

__Note__: Năm 2014 là năm xảy ra nhiều các vụ tấn công khủng bố nhất nên số lượng thương vong lớn nhất là điều dễ hiểu. Bên cạnh đó, trong khoảng thời gian đầu những năm 2000, ta thấy số lượng thương vong năm 2001 tăng vượt bậc so với những năm gần đó, vậy lí do của sự tăng đột biến này là gì?

### Vụ khủng bố 11/09 năm 2001 tại Hoa Kì

Phân tích số liệu trong tập dữ liệu sau đó đối chiếu thực tế.

```python
#Xác định nhóm khủng bố nào gây ra nhiều thương vong nhất ở Hoa Kì vào ngày xảy ra sự kiện 11/09

#lọc các vụ khủng bố xảy ra ở Hoa Kì vào ngày 11/09/2001
df_11_9_2001 = df[(df['iday'] == 11) & (df['imonth'] == 9) & (df['iyear'] == 2001) & (df['country_txt'] == 'United States')]
  
#tính tổng số người tử vong gây ra bởi từng nhóm ở Hoa Kì vào ngày 11/09/2001
number_of_kill_2001 = df_11_9_2001.groupby('gname')['nkill'].sum()
number_of_kill_2001

#tính tổng số người bị thương gây ra bởi từng nhóm ở Hoa Kì vào ngày 11/09/2001
number_of_wound_2001 = df_11_9_2001.groupby('gname')['nwound'].sum()
number_of_wound_2001
```

__Note__: Trong tập dữ liệu, cuộc khủng bố xảy ra vào ngày 11/09/2001 đã gây ra thương vong nặng nề, cướp đi 3001 sinh mạng và khiến cho 16493 người bị thương. Kết quả tìm kiếm ra duy nhất một tổ chức khủng bố đó là Al-Qaida hay nói cách khác tổ chức này chính là thủ phạm của vụ tấn công khủng bố ngày 11/09.

```python
#Từ tập dữ liệu, xác định số người tử vong của từng thành phố ở Hoa Kì vào ngày 11/09/2001

#tính số người tử vong theo từng thành phố ở Hoa Kì vào ngày 11/09/2001
deaths_by_city_2001 = df_11_9_2001.groupby('city')['nkill'].sum()
deaths_by_city_2001 = deaths_by_city_2001.sort_values(ascending=False)
print(deaths_by_city_2001)

#tạo biểu đồ cột
plt.figure(figsize=(12, 6))
sns.barplot(x=deaths_by_city_2001.index, y=deaths_by_city_2001.values, palette = 'rocket')

#thêm thông tin về biểu đồ
plt.xlabel('City', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.ylabel('Number of deaths', fontsize=12)
plt.title('Number of deaths in the terrorist attacks of November 9, 2001 by city in United States', fontsize=16)

#thêm giá trị lên các cột
for i, v in enumerate(deaths_by_city_2001.values):
    plt.text(i, v, f'{v:,.0f}', ha='center', va='bottom')

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

__Note__: Sự kiện 11 tháng 9 gồm một loạt 4 cuộc tấn công khủng bố có sự tham gia của tổ chức khủng bố Hồi giáo Al-Qaeda nhằm chống lại Hoa Kì diễn ra vào ngày 11 tháng 9 năm 2001.
